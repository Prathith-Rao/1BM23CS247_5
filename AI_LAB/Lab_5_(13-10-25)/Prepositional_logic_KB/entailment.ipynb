```python
from sympy.logic.boolalg import Or, And, Not
from sympy.abc import A, B, C

def generate_truth_assignments(variables):
    """
    Generate all possible truth assignments for the given list of variables.
    Returns a list of dictionaries with each dictionary representing an assignment.
    """
    assignments = []
    n = len(variables)
    for i in range(2 ** n):
        assignment = {}
        for j in range(n):
            assignment[variables[j]] = bool(i & (1 << j))
        assignments.append(assignment)
    return assignments

def evaluate_formula(formula, assignment):
    """
    Evaluate a logical formula under a specific truth assignment.
    """
    return formula.subs(assignment)

def check_entailment(kb, query):
    """
    Check if KB entails Q using truth tables.
    
    Arguments:
    kb -- Knowledge base (list of formulas)
    query -- The query formula
    
    Returns:
    True if KB entails Q, False otherwise.
    """
    # Step 1: Negate the query to get ¬Q
    negated_query = Not(query)
    
    # Step 2: Construct KB' = KB ∪ {¬Q}
    kb_prime = And(*kb)  # Conjunction of all KB formulas
    kb_prime_with_neg_query = And(kb_prime, negated_query)
    
    # Step 3: Generate all possible truth assignments for the variables in KB and Q
    variables = [A, B, C]  # Define the list of variables you are using in KB and query
    assignments = generate_truth_assignments(variables)
    
    # Step 4: Evaluate KB' under all truth assignments
    unsatisfiable = True  # Assume KB' is unsatisfiable until proven otherwise
    satisfying_assignments = []
    
    for assignment in assignments:
        kb_value = evaluate_formula(kb_prime_with_neg_query, assignment)
        
        # If KB' is true for this assignment, the query is not entailed
        if kb_value:
            satisfying_assignments.append(assignment)
            unsatisfiable = False
    
    # Step 5: Provide a detailed explanation
    print("\nKnowledge Base (KB):")
    for formula in kb:
        print(f"  {formula}")
    
    print("\nNegated Query (¬Q):")
    print(f"  {negated_query}")
    
    if unsatisfiable:
        print("\nKB entails Q because KB' (KB ∪ {¬Q}) is unsatisfiable under all truth assignments.")
        return True
    else:
        print("\nKB does not entail Q because KB' is satisfiable for the following truth assignments:")
        for assignment in satisfying_assignments:
            print(f"  {assignment}")
        return False

def main():
    # Example Knowledge Base KB = {A ∨ B, ¬A ∨ C, B ∨ ¬C}
    kb = [Or(A, B), Or(Not(A), C), Or(B, Not(C))]
    
    # Example Query Q = A ∨ C
    query = Or(A, C)
    
    # Check entailment and print the detailed reason
    result = check_entailment(kb, query)
    print(f"\nEntailment result: {result}")

if __name__ == "__main__":
    main()
```
