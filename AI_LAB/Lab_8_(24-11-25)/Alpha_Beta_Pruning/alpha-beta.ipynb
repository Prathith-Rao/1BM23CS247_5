```python
# ----------------------------------------
# ALPHA-BETA PRUNING IMPLEMENTATION
# ----------------------------------------

def alphabeta(node, depth, alpha, beta, maximizingPlayer, get_children, evaluate):
    """
    Alpha-Beta pruning algorithm.

    node               - current game state
    depth              - remaining depth to search
    alpha              - best score for maximizing player so far
    beta               - best score for minimizing player so far
    maximizingPlayer   - True if MAX's turn, False for MIN
    get_children       - function returning all possible next states
    evaluate           - function to compute heuristic value of terminal state
    """

    # Base case: leaf node or depth limit
    if depth == 0 or not get_children(node):
        return evaluate(node)

    if maximizingPlayer:
        maxEval = float("-inf")
        for child in get_children(node):
            eval = alphabeta(child, depth - 1, alpha, beta, False, get_children, evaluate)
            maxEval = max(maxEval, eval)
            alpha = max(alpha, eval)

            # Pruning
            if beta <= alpha:
                break
        return maxEval

    else:  # Minimizing player
        minEval = float("inf")
        for child in get_children(node):
            eval = alphabeta(child, depth - 1, alpha, beta, True, get_children, evaluate)
            minEval = min(minEval, eval)
            beta = min(beta, eval)

            # Pruning
            if beta <= alpha:
                break
        return minEval


# ----------------------------------------
# EXAMPLE: Simple Game Tree
# ----------------------------------------

# This example tree is from typical AI textbooks:
# MAX → MIN → MAX … with leaf values

tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E', 'F'],
    'C': ['G', 'H'],
    'D': [], 'E': [], 'F': [],
    'G': [], 'H': []
}

# Leaf node values
values = {
    'D': 3,
    'E': 5,
    'F': 6,
    'G': 9,
    'H': 1
}

def get_children(node):
    return tree.get(node, [])

def evaluate(node):
    return values.get(node, 0)


# ----------------------------------------
# RUN ALPHA-BETA SEARCH
# ----------------------------------------

result = alphabeta('A', depth=10,
                   alpha=float("-inf"),
                   beta=float("inf"),
                   maximizingPlayer=True,
                   get_children=get_children,
                   evaluate=evaluate)

print("Best minimax value =", result)
```
